## **Daily Challenge: Data Handling and Analysis in Python**

```python
import pandas as pd
df = pd.read_csv('datascience_salaries.csv')
df
```

```python
blank_salary = df['salary'].isna().any()
print('Presence of blank values within "salary" column:', blank_salary)
```

```python
from sklearn.preprocessing import MinMaxScaler
salary_scaler = MinMaxScaler()
df['salary'] = salary_scaler.fit_transform(df[['salary']])
df[['salary']].head()
```

```python
df.to_csv('normalized_salary.csv', index=False)
df
```

**Applying Principal Component Analysis (PCA) for Dimensionality Reduciton**

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
X = df.select_dtypes(include=['int64', 'float64']) # Selecting numeric columns only
X_scaled = StandardScaler().fit_transform(X) # Standardizing data
pca = PCA(n_components=2) #  Application of PCA
reduced_data = pca.fit_transform(X_scaled)
reduced_data
```

**Data Aggregation**

```python
result = df.groupby('experience_level')['salary'].agg(['mean', 'median'])
print(result)
```
